# Tiner Remote OCR — Colab worker

Máy local đang chạy `coordinator.py` và in ra **URL** + **TOKEN**.
Dán hai giá trị đó vào ô bên dưới, chọn GPU (Runtime → Change runtime type →
T4 GPU), rồi chạy lần lượt 2 cell.

Worker sẽ tự xin job, tải PDF, quét và trả markdown về máy local.


```python
#@title 1) Cài đặt engine { display-mode: "form" }
ENGINE = "mineru4" #@param ["mineru4", "paddle", "paddle_structure", "docling", "marker", "dots", "unlimited", "mock"]

import subprocess, sys, urllib.request
print(subprocess.run(["nvidia-smi", "--query-gpu=name,memory.total",
                      "--format=csv,noheader"], capture_output=True, text=True).stdout)
urllib.request.urlretrieve(
    "https://raw.githubusercontent.com/chipmax/Tiner-batch-ocr/master/remote/worker.py",
    "worker.py")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "pymupdf", "requests"], check=True)
subprocess.run([sys.executable, "worker.py", "--install", "--engine", ENGINE], check=True)
print("Cài xong:", ENGINE)
```

```python
#@title 2) Kết nối coordinator và chạy { display-mode: "form" }
COORD_URL = "" #@param {type:"string"}
COORD_TOKEN = "" #@param {type:"string"}
ENGINE = "mineru4" #@param ["mineru4", "paddle", "paddle_structure", "docling", "marker", "dots", "unlimited", "mock"]
TIER = "standard" #@param ["flash", "basic", "standard", "advanced"]
MINUTES = 300 #@param {type:"integer"}
DEVICE = "gpu:0"

import subprocess, sys
assert COORD_URL and COORD_TOKEN, "Hãy dán COORD_URL và COORD_TOKEN từ máy local"
cmd = [sys.executable, "worker.py", "--coord", COORD_URL, "--token", COORD_TOKEN,
       "--engine", ENGINE, "--tier", TIER, "--device", DEVICE,
       "--minutes", str(MINUTES)]
print(" ".join(cmd))
subprocess.run(cmd)
```
